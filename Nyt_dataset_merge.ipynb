```python
import pandas as pd
```

```python
f1 = pd.read_csv('open_library_x_nyt_0_615.csv')
f2 = pd.read_csv('open_library_x_nyt_2.csv')
f3 = pd.read_csv('open_library_x_nyt.csv')
f4 = pd.read_csv('open_library_x_nyt_polina.csv')
f5 = pd.read_csv('open_library_x_nyt_katya.csv')
f6 = pd.read_csv('open_library_x_nyt_eugene.csv') 
f7 = pd.read_csv('open_library_x_nyt_eugene_2.csv') 
```

```python
df = pd.concat(
    [f1, f2, f3, f4, f5, f6, f7],
    ignore_index=True
)
df
```

```python
df_sorted = df.sort_values(by = 'avg_rating')
df_sorted
```

```python
df_sorted_clean = df_sorted.drop_duplicates(keep = 'first', subset = ['title', 'author'])
df_sorted_clean
```
