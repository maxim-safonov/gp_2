```python
from dotenv import load_dotenv

load_dotenv()
```

```python
import os

NYT_API_KEY = os.environ.get("NYT_API_KEY")
NYT_URL = os.environ.get("NYT_URL")
```

```python
import requests
import pandas as pd
```

```python
page = requests.get(f"{NYT_URL}/lists/overview.json", params={"api-key": NYT_API_KEY, "published_date": "2026-01-01"})
res = page.json()
res
```

```python
from datetime import date, timedelta
d = date(2020, 1, 20)
d_arr = []

for i in range(80):
    d_arr.append(str(d))
    d = d + timedelta(28)
d_arr
```

```python
import time

books = []
for i in d_arr:
    result = requests.get(f"{NYT_URL}/lists/overview.json", params={"api-key": NYT_API_KEY, "published_date": i}).json()
    for j in result['results']['lists']:
        books.append(j['books'])
    time.sleep(12)
```

```python
books_all = [b for list_item in books for b in list_item]
len(books_all)
```

```python
df = pd.DataFrame(books_all)
df.to_excel("nyt_books.xlsx")
df.head()
```
